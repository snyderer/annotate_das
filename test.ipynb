```python
import numpy as np
import matplotlib.pyplot as plt

c = 1500

src_loc = [500,0,0]
seg = [
    np.arange(0,1000, 10),
    np.zeros((100)),
    np.arange(200, 300, 1)
]

seg = np.asanyarray(seg).T


rng_dir = np.linalg.norm(seg - src_loc, axis=1)
t_dir = rng_dir / c

plt.figure(figsize=(10, 5))    
plt.plot(t_dir)
plt.plot(rng_dir)
plt.show()
```
